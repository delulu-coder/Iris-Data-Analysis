#**Iris Data Analysis**


**Import libararies**


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#for overall appearance of project
sns.set(style="whitegrid")
```

**Load Dataset**

```python
df = sns.load_dataset('iris')
```

**Data Inspection**

```python
print("Shape of dataset:", df.shape)
print("\nColumns:", df.columns)

df.head()
df.info()
df.describe()
```

**📊 Visualization**

**1.Scatter Plot (Core Insights)**

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species')
plt.title("Sepal Length vs Sepal Width")
plt.show()
```

Some species cluster clearly → classification is possible

**2.Pairplot**

```python
sns.pairplot(df, hue='species')
plt.show()
```

**🔹 3. Histograms**

```python
df.hist(figsize=(10,8))
plt.suptitle("Feature Distributions", fontsize=16)
plt.show()
```

**🔹 4. Box Plot**


```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df)
plt.title("Boxplot for Outlier Detection")
plt.show()
```

**Correlation Heatmap**

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Feature Correlation")
plt.show()
```

**Conclusion**


Key Observations:
- Dataset is clean with no missing values
- Petal features are more useful for classification
- Clear clustering between species in scatter plots
- Strong correlation between petal length and width

Conclusion:
The dataset is well-structured and suitable for classification tasks.
